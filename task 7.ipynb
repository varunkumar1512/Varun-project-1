```python
season_order = [
    "Kharif",
    "Rabi",
    "Whole Year",
    "Summer",
    "Autumn",
    "Winter"
]
```

```python
ordinal = OrdinalEncoder(
    categories=[season_order],
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

df["Season_Ordinal"] = ordinal.fit_transform(
    df[["Season"]]
)

df[["Season", "Season_Ordinal"]].head()
```

```python
# Standard deviation and variance together
for col in num_cols:

    print("\nColumn:", col)

    print("Standard Deviation:", df[col].std())

    print("Variance:", df[col].var())
```

```python
#skewnewss
df[num_cols].skew()
```

```python
for col in num_cols:

    skew_value = df[col].skew()

    print("\nColumn:", col)

    print("Skewness:", skew_value)

    if skew_value > 0:
        print("Distribution: Positive Skew")

    elif skew_value < 0:
        print("Distribution: Negative Skew")

    else:
        print("Distribution: Symmetric")
```

```python
#kurtosis
df[num_cols].kurt()
```

```python
for col in num_cols:

    kurt_value = df[col].kurt()

    print("\nColumn:", col)

    print("Kurtosis:", kurt_value)
```

```python
for col in num_cols:

    mean_value = df[col].mean()

    std_value = df[col].std()

    CV = (std_value / mean_value) * 100

    print("\nColumn:", col)

    print("Mean:", mean_value)

    print("Standard Deviation:", std_value)

    print("Coefficient of Variation:", CV, "%")
```

```python
for col in num_cols:

    mean_value = df[col].mean()

    median_value = df[col].median()

    print("\nColumn:", col)

    print("Mean:", mean_value)

    print("Median:", median_value)

    if mean_value > median_value:
        print("Distribution: Positive Skew")

    elif mean_value < median_value:
        print("Distribution: Negative Skew")

    else:
        print("Distribution: Approximately Symmetric")
```
